```python
from datasets import load_dataset
from sklearn.model_selection import train_test_split
import pandas as pd

# Load the dataset
dataset = load_dataset("tuandunghcmut/workflowllm-bench")

# Access splits
train_data_dataset = dataset["train"]  # if available

# Convert to pandas DataFrame for train_test_split compatibility
train_data_df = train_data_dataset.to_pandas()

# Example: Reduce the dataset to 50%
# Splitting the DataFrame to get 50% of the data. We'll take the first split.
reduced_train_data_df, _ = train_test_split(train_data_df, test_size=0.5, random_state=42)

# Display the head of the 50% reduced DataFrame
print("50% Reduced DataFrame Head:")
display(reduced_train_data_df.head())
```

```python
reduced_train_data_df.to_csv("workflowLLM_data.csv")
```
